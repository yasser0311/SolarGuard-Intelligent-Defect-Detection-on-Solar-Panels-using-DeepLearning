# ***✅ Step 1: Extract and Inspect the Dataset***

````python
import zipfile
import os

# Define paths
zip_path = "D:\\Projects\\Guvi_Project5\\SolarGuard\\Faulty_solar_panel-20250430T004242Z-001.zip"
extract_path = "D:\\Projects\\Guvi_Project5\\SolarGuard\\data\\faulty_solar_panel_dataset"

# Extract the zip file
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# Walk through the extracted directory and list the structure
dir_structure = {}
for root, dirs, files in os.walk(extract_path):
    relative_root = os.path.relpath(root, extract_path)
    dir_structure[relative_root] = {
        "dirs": dirs,
        "files": [f for f in files if f.lower().endswith(('.jpg', '.png', '.jpeg'))]
    }

dir_structure
````

````python
import shutil
from PIL import Image

# Output directory for processed images
processed_dir = "D:\\Projects\\Guvi_Project5\\SolarGuard\\data\\processed"
os.makedirs(processed_dir, exist_ok=True)

# Target image size
img_size = (224, 224)

def normalize_and_save_image(src_path, dst_path):
    try:
        img = Image.open(src_path).convert("RGB")
        img = img.resize(img_size)
        img.save(dst_path, format="JPEG")
    except Exception as e:
        print(f"Failed to process {src_path}: {e}")

# Process each class folder
src_base = os.path.join(extract_path, "Faulty_solar_panel")

for class_folder in os.listdir(src_base):
    class_path = os.path.join(src_base, class_folder)
    if not os.path.isdir(class_path):
        continue

    dst_class_path = os.path.join(processed_dir, class_folder.replace(" ", "_"))
    os.makedirs(dst_class_path, exist_ok=True)

    for root, _, files in os.walk(class_path):
        for file in files:
            if file.lower().endswith(('.jpg', '.jpeg', '.png')):
                src_file = os.path.join(root, file)
                dst_file = os.path.join(dst_class_path, os.path.splitext(file)[0] + ".jpg")
                normalize_and_save_image(src_file, dst_file)

# Verify result
processed_summary = {
    cls: len(files)
    for cls in os.listdir(processed_dir)
    if (files := os.listdir(os.path.join(processed_dir, cls)))
}

processed_summary
````

# ***✅ Step 2: Train/Test Split + Data Augmentation***

````python
import random
from shutil import copy2

# Define paths
final_base = "D:\\Projects\\Guvi_Project5\\SolarGuard\\data\\final"
train_dir = os.path.join(final_base, "train")
val_dir = os.path.join(final_base, "val")
os.makedirs(train_dir, exist_ok=True)
os.makedirs(val_dir, exist_ok=True)

split_ratio = 0.8  # 80% training, 20% validation

# Create train/val split for each class
for class_name in os.listdir(processed_dir):
    class_path = os.path.join(processed_dir, class_name)
    images = [f for f in os.listdir(class_path) if f.endswith(".jpg")]
    random.shuffle(images)

    split_idx = int(len(images) * split_ratio)
    train_imgs, val_imgs = images[:split_idx], images[split_idx:]

    # Create class folders
    train_class_dir = os.path.join(train_dir, class_name)
    val_class_dir = os.path.join(val_dir, class_name)
    os.makedirs(train_class_dir, exist_ok=True)
    os.makedirs(val_class_dir, exist_ok=True)

    # Copy images
    for img in train_imgs:
        copy2(os.path.join(class_path, img), os.path.join(train_class_dir, img))
    for img in val_imgs:
        copy2(os.path.join(class_path, img), os.path.join(val_class_dir, img))

# Summary of split
train_summary = {cls: len(os.listdir(os.path.join(train_dir, cls))) for cls in os.listdir(train_dir)}
val_summary = {cls: len(os.listdir(os.path.join(val_dir, cls))) for cls in os.listdir(val_dir)}

train_summary, val_summary
````

````python
import zipfile
import os

# Redefine new zip and extract path
zip_path = "D:\\Projects\\Guvi_Project5\\SolarGuard\\Object Detection.v3i.yolov5pytorch.zip"
extract_path = "D:\\Projects\\Guvi_Project5\\SolarGuard\\data\\final_yolo_dataset"

# Extract the new ZIP
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# Check for data.yaml and class list
yaml_path = os.path.join(extract_path, "data.yaml")
yaml_exists = os.path.exists(yaml_path)

# Parse class names from YAML
classes = []
if yaml_exists:
    with open(yaml_path, 'r') as f:
        for line in f:
            if line.strip().startswith('names:'):
                class_list_line = line.strip()
                break
        else:
            class_list_line = None
    if class_list_line:
        try:
            classes = eval(class_list_line.split(":", 1)[1].strip())
        except Exception:
            pass

# Structure summary
structure = {}
for split in ["train", "valid", "test"]:
    images_dir = os.path.join(extract_path, split, "images")
    labels_dir = os.path.join(extract_path, split, "labels")
    images = os.listdir(images_dir) if os.path.exists(images_dir) else []
    labels = os.listdir(labels_dir) if os.path.exists(labels_dir) else []
    structure[split] = {
        "images": len(images),
        "labels": len(labels)
    }

{
    "yaml_found": yaml_exists,
    "classes": classes,
    "structure_summary": structure
}
````
